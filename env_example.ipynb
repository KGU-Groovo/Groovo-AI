## 환경변수 설정 (.env)

아래 값을 채워서 프로젝트 루트에 `.env` 파일로 저장하세요.  
`JWT_SECRET`은 Spring Boot와 **반드시 동일한 값**이어야 합니다.

| 키 | 설명 |
|---|---|
| `REDIS_URL` | Redis 연결 주소 |
| `JWT_SECRET` | Spring Boot와 공유하는 JWT 시크릿 |
| `AWS_ACCESS_KEY_ID` | S3 접근용 IAM 키 |
| `AWS_SECRET_ACCESS_KEY` | S3 접근용 IAM 시크릿 |
| `AWS_REGION` | S3 버킷 리전 |
| `S3_BUCKET_NAME` | keypoint .npy 파일이 저장된 버킷 이름 |

```python
env_content = """REDIS_URL=redis://localhost:6379
JWT_SECRET=your-jwt-secret-shared-with-spring-boot
AWS_ACCESS_KEY_ID=your-access-key
AWS_SECRET_ACCESS_KEY=your-secret-key
AWS_REGION=ap-northeast-2
S3_BUCKET_NAME=groovo-keypoints
"""

with open(".env", "w") as f:
    f.write(env_content)

print(".env 생성 완료")
```
